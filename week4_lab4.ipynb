## Lab 4.1: Student Management System (+ Challenge เมนู)

```python
# lab4_1.py
# Lab 4.1: Student Management System (List Operations)

# 1. สร้าง List ว่างเปล่า
student_names = []

# ---------------------------------------------------------------
# ส่วนที่ 1: ทำตามขั้นตอนที่ 2-6 ตามลำดับ
# ---------------------------------------------------------------

# 2. เพิ่มรายชื่อนักเรียน (กรอก 3-5 คน)
print("=== 2. Add Students ===")
count = 0
while count < 5:
    name = input(f"Enter student name #{count + 1} (กด Enter เว้นว่างเพื่อหยุด เมื่อครบ 3 คนแล้ว): ").strip()
    if name == "":
        if count >= 3:
            break
        print("ต้องกรอกอย่างน้อย 3 คน")
        continue
    student_names.append(name)
    count += 1
print("Current list:", student_names)

# 3. ค้นหารายชื่อนักเรียน
print("\n=== 3. Find a Student ===")
target = input("Enter a name to search: ").strip()
if target in student_names:
    print(f"Found '{target}' at index {student_names.index(target)}")
else:
    print(f"'{target}' not found")

# 4. ลบรายชื่อนักเรียน
print("\n=== 4. Remove a Student ===")
target = input("Enter a name to remove: ").strip()
if target in student_names:
    student_names.remove(target)
    print(f"Removed '{target}'")
else:
    print(f"'{target}' not found, nothing removed")
print("Updated list:", student_names)

# 5. จัดเรียงรายชื่อ
print("\n=== 5. Sort Students ===")
student_names.sort()
print("Sorted list:", student_names)

# 6. นับจำนวนนักเรียน
print("\n=== 6. Count Students ===")
print("Total students:", len(student_names))


# ---------------------------------------------------------------
# Challenge: ระบบเมนูด้วย while loop
# ---------------------------------------------------------------
print("\n=== Challenge: Menu System ===")
students = []  # เริ่มรายชื่อใหม่สำหรับระบบเมนู

while True:
    print("\n1. เพิ่ม")
    print("2. ลบ")
    print("3. ค้นหา")
    print("4. เรียงลำดับ")
    print("5. ออกจากโปรแกรม")
    choice = input("เลือกเมนู (1-5): ").strip()

    if choice == "1":
        name = input("ชื่อที่ต้องการเพิ่ม: ").strip()
        if name:
            students.append(name)
            print(f"เพิ่ม '{name}' แล้ว")
        else:
            print("ชื่อห้ามว่าง")
    elif choice == "2":
        name = input("ชื่อที่ต้องการลบ: ").strip()
        if name in students:
            students.remove(name)
            print(f"ลบ '{name}' แล้ว")
        else:
            print(f"ไม่พบ '{name}'")
    elif choice == "3":
        name = input("ชื่อที่ต้องการค้นหา: ").strip()
        if name in students:
            print(f"พบ '{name}' ที่ตำแหน่ง index {students.index(name)}")
        else:
            print(f"ไม่พบ '{name}'")
    elif choice == "4":
        students.sort()
        print("เรียงลำดับแล้ว")
    elif choice == "5":
        print("ออกจากโปรแกรม")
        break
    else:
        print("กรุณาเลือก 1-5 เท่านั้น")
        continue

    print("รายชื่อปัจจุบัน:", students, "| จำนวน:", len(students))
```

## Lab 4.2: Geometric Calculations (+ Challenge)

```python
# lab4_2.py
# Lab 4.2: Geometric Calculations (Tuples & Immutability)
import math

# 1. กำหนดจุดเป็น Tuple
point1 = (3, 4)
point2 = (6, 8)

# เข้าถึงและพิมพ์พิกัด x, y
print(f"Point 1: x = {point1[0]}, y = {point1[1]}")
print(f"Point 2: x = {point2[0]}, y = {point2[1]}")

# 2. ทดลองแก้ไขค่า (พิสูจน์ Immutability)
# point1[0] = 5
# ผลที่ได้: TypeError: 'tuple' object does not support item assignment
# เหตุผล: Tuple เป็นข้อมูลชนิด immutable สร้างแล้วเปลี่ยนค่าสมาชิกไม่ได้
#         (ต่างจาก list ที่เป็น mutable) จึงเหมาะกับข้อมูลที่ไม่ควรเปลี่ยน เช่น พิกัด

# 3. คำนวณระยะห่าง: sqrt((x2-x1)^2 + (y2-y1)^2)
x1, y1 = point1
x2, y2 = point2
distance = math.sqrt((x2 - x1) ** 2 + (y2 - y1) ** 2)
print(f"Distance between point1 and point2 = {distance}")  # 5.0

# ---------------------------------------------------------------
# Challenge: รับค่าพิกัดจากผู้ใช้
# ---------------------------------------------------------------
print("\n=== Challenge: Enter your own points ===")


def read_point(label):
    """อ่านพิกัด x, y จาก input() แปลงเป็น float แล้วคืนค่าเป็น tuple"""
    while True:
        try:
            x = float(input(f"{label} x: "))
            y = float(input(f"{label} y: "))
            return (x, y)
        except ValueError:
            print("กรุณากรอกเป็นตัวเลข")


new_point1 = read_point("Point 1")
new_point2 = read_point("Point 2")
new_distance = math.sqrt((new_point2[0] - new_point1[0]) ** 2
                         + (new_point2[1] - new_point1[1]) ** 2)
print(f"Distance between {new_point1} and {new_point2} = {new_distance:.2f}")
```

# Activity 4.1: List vs. Tuple: Scenario Matching

| # | สถานการณ์ | เลือก | เหตุผล |
|---|-----------|-------|--------|
| 1 | อุณหภูมิประจำวันที่บันทึกตลอดหนึ่งสัปดาห์ | **List** | เป็นข้อมูลที่เพิ่มทีละวันและอาจแก้ค่าที่บันทึกผิดได้ ต้องการ `append()` / แก้ค่า และใช้ `sort()` / `len()` วิเคราะห์ต่อได้ |
| 2 | ค่าสี RGB เช่น `(255, 0, 0)` | **Tuple** | เป็นชุดค่า 3 ตัวที่ตายตัว (R, G, B) ค่าสีหนึ่งเฉดไม่ควรถูกแก้โดยไม่ตั้งใจ และใช้เป็น key ของ dictionary ได้ |
| 3 | สินค้าในรถเข็นที่เพิ่ม/ลบได้ | **List** | ต้องเปลี่ยนแปลงตลอด ใช้ `append()` เพิ่ม และ `remove()` / `pop()` ลบ ซึ่ง tuple ทำไม่ได้ |
| 4 | ชื่อวันในสัปดาห์ (จันทร์-อาทิตย์) | **Tuple** | มี 7 วันเสมอ ลำดับคงที่ ไม่มีใครเพิ่มหรือลบวัน ใช้ tuple กันการแก้ไขโดยไม่ตั้งใจ |
| 5 | คะแนนเก็บจากงานชุดต่างๆ ที่อาจแก้ไขภายหลัง | **List** | โจทย์ระบุว่าอัปเดตได้ ต้องเพิ่มคะแนนงานใหม่และแก้คะแนนเดิม (`scores[i] = ...`) ซึ่งต้องใช้ข้อมูลแบบ mutable |
| 6 | พิกัด GPS ของแลนด์มาร์กที่ตั้งอยู่กับที่ | **Tuple** | สถานที่ไม่ย้าย ค่า (latitude, longitude) จึงไม่ควรเปลี่ยน และใช้เป็น key ของ dictionary ได้ |

**สรุป:** ข้อมูลที่ต้องเพิ่ม ลบ หรือแก้ไข (ข้อ 1, 3, 5) ใช้ **list** ส่วนข้อมูลที่คงที่ตามธรรมชาติ (ข้อ 2, 4, 6) ใช้ **tuple**


## Activity 4.2: Interactive To-Do List

```python
# activity4_2.py
# Activity 4.2: Interactive To-Do List (using List Operations)

todo_list = []

while True:
    print("\n=== To-Do List ===")
    print("1. Add a task")
    print("2. View tasks")
    print("3. Mark task as complete")
    print("4. Exit")
    choice = input("Choose (1-4): ").strip()

    if choice == "1":
        task = input("Task: ").strip()
        if task:
            todo_list.append(task)
            print(f"Added '{task}'")
        else:
            print("Task cannot be empty")
    elif choice == "2":
        if len(todo_list) == 0:
            print("No tasks yet")
        else:
            print(f"You have {len(todo_list)} task(s):")
            for i in range(len(todo_list)):
                print(f"  {i + 1}. {todo_list[i]}")
    elif choice == "3":
        if len(todo_list) == 0:
            print("No tasks to complete")
        else:
            for i in range(len(todo_list)):
                print(f"  {i + 1}. {todo_list[i]}")
            number = input("Enter task number or name to complete: ").strip()
            if number.isdigit() and 1 <= int(number) <= len(todo_list):
                done = todo_list.pop(int(number) - 1)
                print(f"Completed '{done}'")
            elif number in todo_list:
                todo_list.remove(number)
                print(f"Completed '{number}'")
            else:
                print("Task not found")
    elif choice == "4":
        print("Goodbye!")
        break
    else:
        print("Please choose 1-4")
```
